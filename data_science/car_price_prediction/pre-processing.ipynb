```python
import pandas as pd 
import numpy as np
```

```python
data_path = './car-price.csv'
data = pd.read_csv(data_path)
initial_frame = pd.DataFrame(data)
print(initial_frame.head(3))
```

```python
# Stats
column_names = initial_frame.columns
print(initial_frame.dtypes)
print('\n\n')
initial_frame.describe()
```

```python
# Dropping columns
p_ds = initial_frame.drop(['car_ID', 'CarName'], axis=1)
```

```python
print(p_ds.head(2))
```

```python
# Categorical & Numerical Columns
num_cols = p_ds.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = p_ds.select_dtypes(include=['object']).columns.tolist()
num_cols.remove('price')

print(f"{num_cols}\n\n {cat_cols}")
```

```python
# Check Null Attr
p_ds.isnull().sum()
```

```python
# further classify categorical columns:

binary_cols = []
multi_cols = []

for col in cat_cols:
    unique_vals = p_ds[col].dropna().unique()
    if len(unique_vals) == 2:
        binary_cols.append(col)
    else:
        multi_cols.append(col)
```

```python
print(f"{binary_cols} \n\n {multi_cols}")
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

le = LabelEncoder()
for col in binary_cols:
    p_ds[col] = le.fit_transform(p_ds[col])

# one-hot enocding
p_ds = pd.get_dummies(p_ds, columns=multi_cols, drop_first=True).astype(int)
```

```python
print(p_ds.head(2))
```

```python
# Visualize the distribution of numerical cols
import seaborn as sns
import matplotlib.pyplot as plt

for col in num_cols:
    sns.histplot(p_ds[col], kde=True)
    plt.show()
#     sns.boxplot(x=p_ds[col])
#     plt.title(f"Boxplot of {col}")
#     plt.show()
```

```python
# Removing Outliers

from scipy import stats

z_scores = stats.zscore(p_ds[num_cols])
abs_Z_score = abs(z_scores)
filtered_entries = (abs_Z_score < 3).all(axis=1)

p_ds = p_ds[filtered_entries]
```

```python
# Scaling / normalization based on plots SC for bell curves & Min-Max for others
skewed_cols = ['symboling', 'boreratio', 'compressionratio']
normal_cols = [col for col in num_cols if col not in skewed_cols]

print(skewed_cols, "\n\n", normal_cols)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

sc = StandardScaler()
mms = MinMaxScaler()

p_ds[normal_cols] = sc.fit_transform(p_ds[normal_cols])
p_ds[skewed_cols] = mms.fit_transform(p_ds[skewed_cols])
```

```python
print(p_ds.describe())
# print(p_ds.head(2))
```

```python
# Save processed dataset to CSV
p_ds.to_csv('p_car_dataset.csv', index=False)
```
