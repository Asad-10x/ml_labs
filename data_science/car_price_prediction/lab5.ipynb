```python
import numpy as np
import pandas as pd
```

```python
path = './p_car_dataset.csv'
df = pd.read_csv(path)
```

```python
from sklearn.feature_selection import SelectKBest, chi2

# Separate features and target
X = df.drop('symboling', axis=1)
y = df['symboling']

# Apply Chi-Square feature selection
chi2_selector = SelectKBest(score_func=chi2, k=10)  # Select top 10 features
X_kbest = chi2_selector.fit_transform(X, y)

# Get selected feature names
selected_features = X.columns[chi2_selector.get_support()]
chi2_scores = chi2_selector.scores_[chi2_selector.get_support()]

# Display results
for feature, score in zip(selected_features, chi2_scores):
    print(f"{feature}: {score:.2f}")
```
